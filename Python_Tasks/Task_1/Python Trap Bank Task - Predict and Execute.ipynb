```python
##Q35
import copy
a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a
```

```python
# [[1, 99], [3, 4]]
# copy.copy() performs a shallow copy. 
# It duplicates the outer list, but the inner list objects are still shared between a and b. Modifying an inner list alters both.
```

```python
import copy
a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
```

```python
## Q36
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
```

```python
# [[1, 2], [3, 4]]
#copy.deepcopy() recursively duplicates everything, including all inner elements. 
# The object sequences are completely decoupled, so modifying b leaves a untouched.
```

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
```

```python
## Q37
import copy

a = [[10], [20]]
b = a.copy()
a[1].append(30)
print(b)
```

```python
# [[10], [20, 30]]
# a.copy() is a shallow copy. The nested lists are references, meaning updating a[1] also modifies b[1].
```

```python
import copy

a = [[10], [20]]
b = a.copy()
a[1].append(30)
print(b)
```

```python
## Q38
import copy

a = [[10], [20]]
b = copy.deepcopy(a)
a[1].append(30)
print(b)
```

```python
# [[10], [20]]
# A deep copy prevents inner elements from sharing references, so appending to a[1] does not affect b.
```

```python
import copy

a = [[10], [20]]
b = copy.deepcopy(a)
a[1].append(30)
print(b)
```
