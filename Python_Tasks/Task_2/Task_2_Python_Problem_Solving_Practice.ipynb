## Q1. Find the Shortest Word in a Sentence

**1. Observe:** Input: `Python makes coding easy`; expected output: `easy`.

**2. Understand:** Return a word with the fewest characters.

**3. Break:** Split the sentence into words; track the shortest word while traversing them.

**4. Logic:** Start with the first word. Replace the current answer only when a shorter word is found, so ties keep the first word.

**5. Code:** Run the following cell.

**6. Test:** Check the given example and an additional test case.

```python
def shortest_word(sentence):
    words = sentence.split()
    if not words:
        raise ValueError("Sentence must contain at least one word")
    shortest = words[0]
    for word in words[1:]:
        if len(word) < len(shortest):
            shortest = word
    return shortest

print(shortest_word("Python makes coding easy"))
```

```python
assert shortest_word("Python makes coding easy") == "easy"
assert shortest_word("red blue tan") == "red"
print(shortest_word("Python makes coding easy"))
print(shortest_word("red blue tan"))
```

**6. Test:** Check the given example and an additional test case.

```python
def largest_number(numbers):
    if not numbers:
        raise ValueError("List must not be empty")
    largest = numbers[0]
    for number in numbers[1:]:
        if number > largest:
            largest = number
    return largest

print(largest_number([12, 45, 7, 89, 23]))
```

## Q6. Count Vowels in a String

**1. Observe:** Input: `Education`; expected output: `5`.

**2. Understand:** Count all occurrences of a, e, i, o, and u in either case.

**3. Break:** Visit each character; normalize its case; count it if it is a vowel.

**4. Logic:** Use a vowel set and a counter, incrementing for each matching character.

**5. Code:** Run the following cell.

**6. Test:** Check the given example and an additional test case.

```python
def count_vowels(text):
    vowels = "aeiou"
    count = 0
    for character in text.lower():
        if character in vowels:
            count += 1
    return count

print(count_vowels("Education"))
```

## Q7. Count Consonants in a String

**1. Observe:** Input: `Python 123!`; expected output: `5`.

**2. Understand:** Count alphabetic letters that are not vowels, ignoring digits, spaces, and symbols.

**3. Break:** Traverse the lowercase text; check if each character is a letter; count it if it is not a vowel.

**4. Logic:** Use `isalpha()` to exclude non-letters and a vowel string to identify consonants.

**5. Code:** Run the following cell.

**6. Test:** Check the given example and an additional test case.

```python
def count_consonants(text):
    vowels = "aeiou"
    count = 0
    for character in text.lower():
        if character.isalpha() and character not in vowels:
            count += 1
    return count

print(count_consonants("Python 123!"))
```

```python
assert count_consonants("Python 123!") == 5
assert count_consonants("AEIOU xyz") == 3
print(count_consonants("Python 123!"))
print(count_consonants("AEIOU xyz"))
```

## Q8. Find the Most Frequent Character

**1. Observe:** Input: `banana`; expected output: `a`.

**2. Understand:** Return the character with the highest frequency, ignoring spaces; ties go to the first encountered.

**3. Break:** Remove spaces; count each character; find the highest count while preserving first appearance order.

**4. Logic:** Build a frequency dictionary in encounter order, then update the result only for a strictly greater count.

**5. Code:** Run the following cell.

**6. Test:** Check the given example and an additional test case.

```python
def most_frequent_character(text):
    text = "".join(text.split())
    if not text:
        raise ValueError("Input must contain a non-space character")
    frequency = {}
    for character in text:
        frequency[character] = frequency.get(character, 0) + 1
    result = text[0]
    for character in frequency:
        if frequency[character] > frequency[result]:
            result = character
    return result

print(most_frequent_character("banana"))
```

```python
assert most_frequent_character("banana") == "a"
assert most_frequent_character("a b b a") == "a"
print(most_frequent_character("banana"))
print(most_frequent_character("a b b a"))
```

## Q18. Find the First Word That Appears Only Once

**1. Observe:** Input: `apple banana apple mango banana orange`; expected output: `mango`.

**2. Understand:** Return the first word whose case-insensitive frequency is exactly one.

**3. Break:** Normalize words to lowercase; count each; traverse the normalized sequence to find the first count of one.

**4. Logic:** Build a frequency dictionary and return the first word with frequency one; raise an error if none exists.

**5. Code:** Run the following cell.

**6. Test:** Check the given example and an additional test case.

```python
def first_unique_word(sentence):
    words = sentence.lower().split()
    frequency = {}
    for word in words:
        frequency[word] = frequency.get(word, 0) + 1
    for word in words:
        if frequency[word] == 1:
            return word
    raise ValueError("No word appears exactly once")

print(first_unique_word("apple banana apple mango banana orange"))
```

```python
assert first_unique_word("apple banana apple mango banana orange") == "mango"
assert first_unique_word("Hi hi THERE") == "there"
print(first_unique_word("apple banana apple mango banana orange"))
print(first_unique_word("Hi hi THERE"))
```

## Q20. Find the Longest Word That Appears Only Once

**1. Observe:** Input: `Python makes coding useful Python`; expected output: `coding`.

**2. Understand:** Find the longest word occurring once, treating uppercase and lowercase as the same; ties go to the first.

**3. Break:** Lowercase and split; count word frequencies; consider only words with count one; track the longest candidate.

**4. Logic:** Traverse words in order and update the answer only for a strictly longer word whose frequency is one. Raise an error if there are no unique words.

**5. Code:** Run the following cell.

**6. Test:** Check the given example and an additional test case.

```python
def longest_unique_word(sentence):
    words = sentence.lower().split()
    frequency = {}
    for word in words:
        frequency[word] = frequency.get(word, 0) + 1

    result = None
    for word in words:
        if frequency[word] == 1:
            if result is None or len(word) > len(result):
                result = word

    if result is None:
        raise ValueError("No word appears exactly once")
    return result

print(longest_unique_word("Python makes coding useful Python"))
```

```python
assert longest_unique_word("Python makes coding useful Python") == "coding"
assert longest_unique_word("one TWO three four") == "three"
print(longest_unique_word("Python makes coding useful Python"))
print(longest_unique_word("one TWO three four"))
```
